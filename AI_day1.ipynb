```python
file=open("sample.txt","r") #We are creating a file sample.txt and open in it write mode
content =file.read() # This is reading all the content in one go
print(content)
file.close()
```

```python
file=open("sample.txt","r") #We are creating a file sample.txt and open in it write mode
for line in file:
  print(line.strip()) # Strip is using to remove leading and trailing spaces
file.close()
```

```python
file=open("sample.txt","a") #We are creating a file sample.txt and open in it write mode
file.write("\n I have added a new line")
file.close()
```

```python
with open("sample.txt","r") as file: #We are creating a file sample.txt and open in it write mode
#  content =file.read() # This is reading all the content in one go
  for line in file:
    print(line.strip())
#  print(content)
# with open automatically closes the file handle
```

```python
with open("sample.txt", "w") as file:
  file.write("hello world\n")
  file.write("hello new world")
```

```python
# exception handling

try: # here we write anything in suspition that it might fail
  with open("sample.txt", "r") as file:
    print(file.read())
except FileNotFoundError: # this is a block where control comes when anything fails in try
  print("my file sample.txt is not found")
```
