# Part 1 — What Python is, and how we run it

## 1.1 What is Python?
Python is a programming language. A programming language is simply a way to give instructions to a computer.

Python is popular because it is readable like English and very powerful for automation, data, web, and AI.


## 1.2 How Python runs
Python code is executed by a program called the **Python Interpreter**.
The interpreter reads code **line by line** and executes it.


## 1.3 Two ways to run Python
1. Interactive mode (like chatting)
2. Script mode (.py file)
