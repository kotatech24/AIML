```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
real_data = np.random.normal(loc = 5000, scale = 500, size = 1000)

# this data tells about customer spend
# np.random.normal => this function generates random numbers from normal distribution (bell shaped distribution)
# loc - means average/ mean of disctribution
# scale - std deviation - how spread out the numbers are
# size = 1000 - generate 1000 numbers
# because mean is 5000 and std dev is 500 so many numbers will be between 4500 and 5500
# as per 68-95-99.7 rule - 68% values are between 4500 and 5500, 95% values are in +- 2 std = 4000 and 6000, 99.7 3500 and 6500

real_data
```

```python
generator_means = []
generator_mean = 1000
generator_std = 2000

for epoch in range(500):
  fake_data = np.random.normal(loc = generator_mean, scale = generator_std, size = 1000)
  real_mean = np.mean(real_data)
  fake_mean = np.mean(fake_data)
  difference = real_mean - fake_mean

  # Generator improves
  generator_mean = generator_mean + difference*0.2
  generator_std = generator_std - 20

  if generator_std<500:
    generator_std = 500

  if generator_mean > 5000:
    generator_mean = 5000

  generator_means.append(generator_mean)
  print("Generator improved Mean ", generator_mean)
  print("Generator improved Std ", generator_std)

```
