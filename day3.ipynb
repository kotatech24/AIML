```python
#%%
# frequency distribution
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
text = "I love NLP and I love python and python is amazing"
token = word_tokenize(text)
fdist = FreqDist(token)
print(type(fdist))
#print(fdist)
for word, freq in fdist.items():
    print(word, " - ", freq)
#print(fdist.most_common)
```

```python
fdist.plot(10)
```

```python
# example of bigram
import nltk
from nltk.util import bigrams
from nltk.tokenize import word_tokenize
text = "i love nlp and python"
token = word_tokenize(text)
bigram_list = list(bigrams(token))
print(bigram_list)
```

```python
# example trigram
import nltk
from nltk.util import trigrams
from nltk.tokenize import word_tokenize
text = "i love nlp and python"
token = word_tokenize(text)
trigram_list = list(trigrams(token))
print(trigram_list)
```

```python
#example of ngram
import nltk
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
text = "i love nlp and python"
token = word_tokenize(text)
ngram_list = list(ngrams(token,4))
print(ngram_list)
```

```python
# Example of PorterStemmer
from nltk.stem import PorterStemmer
ps = PorterStemmer()
words = ["Played", "Player", "Playing", "Plays"]
for w in words :
    print(w,"-> ", ps.stem(w))
```

```python
 #Example of Lancaster

from nltk.stem import LancasterStemmer
ls = LancasterStemmer()

words = ["Played", "Player", "Playing", "Plays"]
for w in words :
    print(w,"-> ", ls.stem(w))
```

```python
# Example of Snowball Stemmer

from nltk.stem import SnowballStemmer
sb = SnowballStemmer("english")
words = ["Played", "Player", "Playing", "Plays"]
for w in words :
    print(w,"-> ", sb.stem(w))
```
