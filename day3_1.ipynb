```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
def lematize(words):
  for word in words:
    print(word, '->', lemmatizer.lemmatize(word))

def lematizeAs(words, how):
  print('Lemmatizing as', how)
  for word in words:
    print(word, '->', lemmatizer.lemmatize(word, how))
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag, word_tokenize

def wordnetpos(tag):
  if tag.startswith('J'):
    return wordnet.ADJ
  elif tag.startswith('V'):
    return wordnet.VERB
  elif tag.startswith('N'):
    return wordnet.NOUN
  elif tag.startswith('R'):
    return wordnet.ADV
  else:
    return wordnet.NOUN

lemmatizer = WordNetLemmatizer()

text = input('Please enter some text: ')
tokens = word_tokenize(text)
tagged = pos_tag(tokens)
lemmatized = [lemmatizer.lemmatize(word, wordnetpos(tag)) for word, tag in tagged]
print ('Original Tokens: ', tokens)
print(tagged)
print ('lemmatized: ', lemmatized)
```

```python
import nltk.stem
from nltk.stem import SnowballStemmer
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag, word_tokenize


def read_file(path):
  file = open(path, 'r')
  data = file.read()
  return data

def remove_stop_words(tokens):
  stop_words = set(stopwords.words('english'))
  filtered_tokens = [token for token in tokens if token.lower() not in stop_words]
  return filtered_tokens

def get_wordnet_pos(tag):
  if tag.startswith('J'):
    return wordnet.ADJ
  elif tag.startswith('V'):
    return wordnet.VERB
  elif tag.startswith('N'):
    return wordnet.NOUN
  elif tag.startswith('R'):
    return wordnet.ADV
  else:
    return wordnet.NOUN

def tag_meaning(tag):
  if tag.startswith('J'):
    return 'Adjective'
  elif tag.startswith('V'):
    return 'Verb'
  elif tag.startswith('N'):
    return 'Noun'
  elif tag.startswith('R'):
    return 'Adverb'
  else:
    return 'Noun'


data = read_file('/content/data.txt')
tokens = word_tokenize(data)
filtered_tokens = set(remove_stop_words(tokens))
tagged = pos_tag(filtered_tokens)

sb = SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()


# print('word\t\t\tpos\t\t\tmeaning\t\t\tstem\t\t\tlemma')
word_list,tag_list,tag_meaning_list,stemmed_word_list,lemmatized_list = [],[],[],[],[]

for word,tag in tagged:
  stemmed_word = sb.stem(word)
  lemmatized = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
  word_list.append(word)
  tag_list.append(tag)
  tag_meaning_list.append(tag_meaning(tag))
  stemmed_word_list.append(stemmed_word)
  lemmatized_list.append(lemmatized)

d = {
    'word': word_list, 'tags': tag_list, 'tag_meanings': tag_meaning_list,
    'stemmed_words': stemmed_word_list, 'lemmatized': lemmatized_list
}
df = pd.DataFrame(d)
print(df)
df.to_csv('/content/output.csv', index=False)

  # print (word, '\t\t', tag,'\t\t\t', tag_meaning(tag),'\t\t\t', stemmed_word,'\t\t\t', lemmatized)

# tagged
```
