```python
from sklearn.feature_extraction.text import CountVectorizer
documents = [
    "i love java",
    "i love python",
    "java and python are both popular"
]

vectorizer = CountVectorizer()
```

```python
# convert to numerical form
X = vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out()) # this will give unique words
```

```python
X.toarray()
```
