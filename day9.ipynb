```python
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
import numpy as np
```

```python
import requests

# URL of a sample image
image_url = 'https://upload.wikimedia.org/wikipedia/commons/thumb/a/a2/Image_of_Puglia.jpg/1200px-Image_of_Puglia.jpg'
image_path = '/content/download.jpeg'

# Download the image
response = requests.get(image_url)
with open(image_path, 'wb') as f:
    f.write(response.content)

print(f"Image downloaded to: {image_path}")
```

```python
image = load_img("/content/download.jpeg", target_size = (224,224))
image = img_to_array(image)
image = np.expand_dims(image, axis = 0)
```

```python
datagen = ImageDataGenerator(rotation_range = 60)
plt.figure(figsize=(10,5))
for i, batch in enumerate(datagen.flow(image, batch_size = 1)):
  plt.subplot(2,3,i+1)
  plt.imshow(batch[0].astype(np.uint8))
  plt.axis("off")
  if i == 5:
    break
plt.suptitle("rotated range 30")
plt.show()
```

```python
# width shift range

datagen = ImageDataGenerator(width_shift_range=0.7, fill_mode="nearest")
plt.figure(figsize=(10,5))
for i, batch in enumerate(datagen.flow(image, batch_size = 1)):
  plt.subplot(2,3,i+1)
  plt.imshow(batch[0].astype(np.uint8))
  plt.axis("off")
  if i == 5:
    break
plt.suptitle("width shift by 0.2")
plt.show()
```

```python
# height shift range

datagen = ImageDataGenerator(height_shift_range=0.7, fill_mode="nearest")
plt.figure(figsize=(10,5))
for i, batch in enumerate(datagen.flow(image, batch_size = 1)):
  plt.subplot(2,3,i+1)
  plt.imshow(batch[0].astype(np.uint8))
  plt.axis("off")
  if i == 5:
    break
plt.suptitle("height shift by 0.2")
plt.show()
```

```python
# shear - skweing image

datagen = ImageDataGenerator(shear_range=50, fill_mode="nearest")
plt.figure(figsize=(10,5))
for i, batch in enumerate(datagen.flow(image, batch_size = 1)):
  plt.subplot(2,3,i+1)
  plt.imshow(batch[0].astype(np.uint8))
  plt.axis("off")
  if i == 5:
    break
plt.suptitle("shear")
plt.show()
```

```python
# shear - skweing image

datagen = ImageDataGenerator(zoom_range=0.7, fill_mode="nearest")
plt.figure(figsize=(10,5))
for i, batch in enumerate(datagen.flow(image, batch_size = 1)):
  plt.subplot(2,3,i+1)
  plt.imshow(batch[0].astype(np.uint8))
  plt.axis("off")
  if i == 5:
    break
plt.suptitle("zoom")
plt.show()
```
